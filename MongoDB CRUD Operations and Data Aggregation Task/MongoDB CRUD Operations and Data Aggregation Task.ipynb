```python
from pymongo import MongoClient

# Connect to MongoDB 
client = MongoClient('mongodb://localhost:27017/')

# Create or access the database 'testdb'
db = client.testdb

# Create or access the collection 'users'
users_collection = db.users
print("Database and Collection setup complete.")
```

Insert Documents

```python
user_data = [
    {"name": "Ali", "email": "ali@example.com", "age": 24},
    {"name": "Ahmed", "email": "ahmed@example.com", "age": 32},
    {"name": "Saher", "email": "saher@example.com", "age": 28},
    {"name": "Walid", "email": "walid@example.com", "age": 19},
    {"name": "Emad", "email": "emad@example.com", "age": 45}
]

# Insert the documents
users_collection.insert_many(user_data)
print("Documents inserted")
```

```python
# Retrieve all documents
all_users = users_collection.find()

# Print the retrieved documents
print("All users in the collection:")
for user in all_users:
    print(user)
    print("-"*60)
```

```python
# Update Ali's age
update_result = users_collection.update_one(
    {"name": "Ali"},  # Filter to find Ali
    {"$set": {"age": 30}}  # Update operation: set age to 30
)

print(f"Matched {update_result.matched_count} document(s) and modified {update_result.modified_count} document(s)")
print("-"*60)
# Verify the update
updated_ali = users_collection.find_one({"name": "Ali"})
print(f"Updated Ali: {updated_ali}")
```

```python
# Delete users under 20
delete_result = users_collection.delete_many({"age": {"$lt": 20}})
print(f"Deleted {delete_result.deleted_count} document(s).")
print("-"*60)
# Verify deletion
remaining_users = users_collection.find()
print("Remaining users after deletion:")
for user in remaining_users:
    print(user)
    print("-"*60)
```

Aggregation Tasks

```python
# Aggregation pipeline for average age
pipeline = [
    {"$group": {"_id": None, "averageAge": {"$avg": "$age"}}}
]

# Execute the aggregation
average_age_result = list(users_collection.aggregate(pipeline))

# Print the result
if average_age_result:
    print(f"Average age of users: {average_age_result[0]['averageAge']}")
else:
    print("No users found to calculate average age.")
```

```python
# Aggregation pipeline to find users older than 25
pipeline = [
    {"$match": {"age": {"$gt": 25}}}
]

# Execute the aggregation
older_users = list(users_collection.aggregate(pipeline))

# Print the results
print("Users older than 25:")
for user in older_users:
    print(user)
    print("-"*60)
```

```python
# Aggregation pipeline to sort users by age (descending)
pipeline = [
    {"$sort": {"age": -1}}  # -1 for descending, 1 for ascending
]

# Execute the aggregation
sorted_users = list(users_collection.aggregate(pipeline))

# Print the sorted users
print("Users sorted by age (descending):")
for user in sorted_users:
    print(user)
    print("-"*60)
```

Data Insights and Analysis

```python
# Aggregation pipeline to group users by age and count
pipeline = [
    {"$group": {"_id": "$age", "count": {"$sum": 1}}},
    {"$sort": {"_id": 1}} #sort by age (ascending)
]

# Execute the aggregation
age_groups = list(users_collection.aggregate(pipeline))

# Print the results
print("Number of users in each age group:")
for group in age_groups:
    print(f"Age {group['_id']}: {group['count']} users")
    print("-"*60)
```

```python
# Find the minimum age
pipeline_min_age = [
     {"$group": {"_id": None, "minAge": {"$min": "$age"}}}
]
min_age_result = list(users_collection.aggregate(pipeline_min_age))

if min_age_result:
    min_age = min_age_result[0]['minAge']
    print(f"The minimum age is: {min_age}")

    #Aggregation to show the youngest groups and sort by user count
    pipeline_youngest_group = [
        {"$match": {"age": min_age}},
        {"$group": {"_id": "$age", "users": {"$push": "$name"}, "count": {"$sum": 1}}},
        {"$sort": {"count": -1}} # Sort by the user count descending

    ]
    youngest_group_result = list(users_collection.aggregate(pipeline_youngest_group))
    print("\nYoungest Age Group(s):")
    for group in youngest_group_result:
        print(f"  Age: {group['_id']}, Users: {group['users']}, Count: {group['count']}")
else:
    print("No users in Collection")
```

```python
pipeline = [
    {"$group": {
        "_id": "$age",  # Group by age
        "count": {"$sum": 1},  # Count users in each age group
        "users": {"$push": "$name"}  # Keep track of user names in each group
    }},
    {"$sort": {"_id": 1, "count": -1}},  # Sort by age (ascending), then count (descending)
    {"$group": {
        "_id": None,  # Group all results together
        "ageGroups": {"$push": {"age": "$_id", "count": "$count", "users": "$users"}},  # Array of age groups
        "averageAge": {"$avg": "$_id"},  # Calculate the overall average age
        "minAge": {"$min": "$_id"} # Calculate minimum Age

    }},
    {"$unwind": "$ageGroups"}, #unwind to prepare for the match (next step)
     {"$match": {"$expr": {"$eq": ["$ageGroups.age", "$minAge"]}}}, #keep only the youngest groups
    {"$project": {
        "_id": 0,  # Exclude the default _id field
        "averageAge": {"$round": ["$averageAge", 2]},  # Round average age to 2 decimal places
        "youngestAgeGroups": {
            "age": "$ageGroups.age",
            "count": "$ageGroups.count",
            "users": "$ageGroups.users"
        }
    }}
]
result = list(users_collection.aggregate(pipeline))

if result:
    print("Combined Aggregation Results:")
    print(f"  Average Age: {result[0]['averageAge']}")

    # iterate, because there could be multiple youngest groups
    for group in result:
      print(f"  Youngest Age Group - Age: {group['youngestAgeGroups']['age']}, Count: {group['youngestAgeGroups']['count']}, Users: {group['youngestAgeGroups']['users']}")

else:
    print("No users in Collection.")
```
