# Data Manipulation with Pandas and NumPy 🤗

```python
#libraries
import pandas as pd
import numpy as np
```

# Creating and Manipulating Pandas DataFrames

```python
# Creating a DataFrame from a list of lists
data_list = [['Ali', 25, 'Engineer'],
 ['Ahmed', 29, 'Doctor'],
 ['Mohamed', 22, 'Student']]
data_list_df=pd.DataFrame(data_list,columns=['Name', 'Age', 'Occupation'])
print("DataFrame from List:\n",data_list_df)
```

```python
# Creating a DataFrame from a dictionary
data_dict = {'Name': ['Ali', 'Ahmed', 'Mohamed'],
 'Age': [25, 29, 22],
 'Occupation': ['Engineer', 'Doctor', 'Student']}
data_dict_df=pd.DataFrame(data_dict)
print("DataFrame from Dictionary:\n",data_dict_df)
```

```python
# Creating a DataFrame from a NumPy array
data_array = np.array([['Ali', 25, 'Engineer'],
 ['Ahmed', 29, 'Doctor'],
 ['Mohamed', 22, 'Student']])
data_array_df=pd.DataFrame(data_array,columns=['Name', 'Age', 'Occupation'])
print("DataFrame from Array:\n",data_array_df)
```

```python
# Selecting data
print("Selecting a single column:\n",data_list_df['Age'])
print("-"*60)
print("\nSelecting the first row:\n", data_dict_df.iloc[0])
print("-"*60)
print("\nSelecting 'Age' where 'Occupation' is 'Doctor':\n", data_array_df.loc[data_array_df['Occupation'] == 'Doctor', 'Age'])
```

```python
# Filtering rows based on condition
print("Filtering rows where 'Age' is greater than 25:\n", data_dict_df[data_dict_df['Age'] > 27])
```

# Handling Missing Data and Merging DataFrames

```python
# Creating a DataFrame with missing data
data_missing = {'Name': ['Ali', 'Ahmed', 'Mohamed'],
 'Age': [25, 29, None],
 'Occupation': ['Engineer', 'Doctor', 'Student']}
data_missing_df=pd.DataFrame(data_missing)
print("DataFrame with missing data:\n", data_missing_df)
```

```python
data_missing_df.isnull().sum()
```

```python
# Handling missing data using fillna()
data_missing_df['Age'] = data_missing_df['Age'].fillna(data_missing_df['Age'].mean())
print("DataFrame after handling missing data:\n", data_missing_df)
```

```python
# Creating a DataFrame with missing data
data_missingg = {'Name': ['Ali', 'Ahmed', 'Mohamed','sayed'],
 'Age': [25, 29, 22 , None],
 'Occupation': ['Engineer', 'Doctor', 'Student','lawyer']}
data_missingg_df=pd.DataFrame(data_missingg)
print("DataFrame with missing data:\n", data_missingg_df)
```

```python
# Handling missing data using dropna()
data_missingg_df = data_missingg_df.dropna()
print("DataFrame after dropping rows with missing data:\n", data_missingg_df)
```

```python
# Creating two DataFrames for merging
data_merge1 = {'Name': ['Ali', 'Ahmed', 'Mohamed'],
 'Age': [25, 29,22]}
data_merge1_df=pd.DataFrame(data_merge1)
print("DataFrame 1:\n", data_merge1_df)
print("-"*60)
#---------------------------------------------------
data_merge2 = {'Name': ['Ali', 'Ahmed', 'Mohamed'],
 'Occupation': ['Engineer', 'Doctor', 'Student']}
data_merge2_df=pd.DataFrame(data_merge2)
print("DataFrame 2:\n", data_merge2_df)
```

```python
# Merging DataFrames
merged_df = pd.merge(data_merge1_df, data_merge2_df, on='Name',how='inner')
print("Merged DataFrame:\n", merged_df)
#------------------------------------------
print("-"*60)
#right join
merged_df_rg = pd.merge(data_merge1_df, data_merge2_df, on='Name',how='right')
print("Merged DataFrame:\n", merged_df_rg)
#------------------------------------------
print("-"*60)
#left join
merged_df_lf = pd.merge(data_merge1_df, data_merge2_df, on='Name',how='left')
print("Merged DataFrame:\n", merged_df_lf)
```

# Combining Pandas and NumPy for Analysis

```python
# Creating a DataFrame with numerical data
numerical_data = {'A': [1, 2, 3],
 'B': [4, 5, 6],
 'C': [7, 8, 9]}
numerical_df=pd.DataFrame(numerical_data)
print("DataFrame with numerical data:\n", numerical_df)
```

```python
# Using NumPy for aggregation
print("Sum of each column:\n", np.sum(numerical_df, axis=0))
print("-"*60)
print("Mean of each row:\n", np.mean(numerical_df, axis=1))
print("-"*60)
print("Max of each column:\n", np.max(numerical_df, axis=0))
print("-"*60)
# Calculate standard deviation using NumPy on the DataFrame
print("Standard deviation of each column:\n", np.std(numerical_df, axis=0))
print("-"*60)
# Calculate median using NumPy on the DataFrame
print("Median of each column:\n", np.median(numerical_df, axis=0))
```

```python
print("Square root of each element:\n", np.sqrt(numerical_df))
print("-"*60)
print("Logarithm of each element:\n", np.log(numerical_df))
```

```python
# Using NumPy for data transformation
def custom_function(x):
  return x * 2 + 1

numerical_df['E'] = numerical_df['B'].apply(custom_function)
print("\nDataFrame with transformed column 'B':\n", numerical_df)
```

```python
# apply transformation on data_dict_df
def age_group(age):
    if age < 25:
        return 'Young'
    elif age < 30:
        return 'Adult'
    else:
        return 'Senior'

data_dict_df['Age_Group'] = data_dict_df['Age'].apply(age_group)
print("DataFrame with new 'Age_Group' column:\n", data_dict_df)

# Example transformation using NumPy: Squaring the 'Age' column
data_dict_df['Age_Squared'] = np.square(data_dict_df['Age'])
print("\nDataFrame with squared 'Age' values:\n", data_dict_df)
```

```python
# Statistical analysis using NumPy
correlation_matrix = np.corrcoef(np.array(list(numerical_data.values())).T) # convert the dictionary values to a NumPy array and then transpose.
print("\nCorrelation matrix:\n", correlation_matrix)
```

```python
# Grouping data and performing aggregation
data_group = {'Category': ['X', 'X', 'Y', 'Y', 'X'],
              'Value': [10, 12, 15, 18, 20]}
df_group = pd.DataFrame(data_group)
print("Data_Group\n", df_group)
print("-"*60)
#---------------------------------------------------------
grouped_data = df_group.groupby('Category')['Value'].sum()
print("\nGrouped data and sum:\n", grouped_data)
print("-"*60)
#---------------------------------------------------------
grouped_data_mean = df_group.groupby('Category')['Value'].mean()
print("\nGrouped data and mean:\n", grouped_data_mean)
```
