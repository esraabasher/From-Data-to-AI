# Basic and Advanced Data Visualizations with Matplotlib 🤗

```python
#libraries
import matplotlib.pyplot as plt
import numpy as np
```

# Basic Line and Scatter Plots

```python
# Data for the plots
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

```python
# Line Plot
plt.figure(figsize=(8, 4))
plt.plot(x, y, color='green', label='sin(x)')
plt.title('Basic Line Plot of sin(x)')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(8, 4))
plt.scatter(x, y, color='yellow', label='sin(x) points')
plt.title('Scatter Plot of sin(x)')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Scatter Plot with different markers and colors
x2 = np.random.rand(50)
y2 = np.random.rand(50)
colors = np.random.rand(50)
sizes = 1000 * np.random.rand(50)

plt.scatter(x2, y2, c=colors, s=sizes, alpha=0.5, cmap='viridis')  # Use a colormap
plt.colorbar() # Show colorbar for reference
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Scatter Plot")
plt.show()
```

# Advanced Plotting with Subplots and Annotations

```python
# Data for the subplots
categories = ['A', 'B', 'C', 'D']
values = [32, 45, 56, 66]
data_hist = np.random.normal(50, 15, 100)
```

```python
# Create subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
#-----------------------------------------------------------
# Bar Plot
ax1.bar(categories, values, color=['blue', 'green', 'red', 'purple'])
ax1.set_title('Bar Plot')
ax1.set_xlabel('Categories')
ax1.set_ylabel('Values')
ax1.grid(True)
#-----------------------------------------------------------
# Histogram
ax2.hist(data_hist, bins=20, color='black', edgecolor='brown')
ax2.set_title('Histogram')
ax2.set_xlabel('Values')
ax2.set_ylabel('Frequency')
ax2.grid(True)
#-----------------------------------------------------------
# Add annotation to the bar plot
ax1.annotate('Highest Value', xy=(3, 66), xytext=(3.5, 70),
             arrowprops=dict(facecolor='black', shrink=0.05))

plt.tight_layout()
plt.show()
```

# 3D Plotting

```python
from mpl_toolkits.mplot3d import Axes3D

# Data for the 3D scatter plot
np.random.seed(42)
x = np.random.standard_normal(100)
y = np.random.standard_normal(100)
z = np.random.standard_normal(100)
#---------------------------------------------
# Create 3D scatter plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z, c='green', marker='o')
#---------------------------------------------
# Customize the plot
ax.set_title('3D Scatter Plot')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
#---------------------------------------------
# Save the plot as an image file
plt.savefig('3d_scatter_plot.png', dpi=300)
plt.show()
```
