# Data Visualizations using Seaborn 🤗

```python
#libraries
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

# Basic Plotting with Seaborn.

```python
# Data for the plots
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

```python
# Line Plot
plt.figure(figsize=(8, 4))
sns.lineplot(x=x, y=y, label='sin(x)')
sns.set_style("darkgrid")
sns.set_palette("pastel")
plt.title('Line Plot of sin(x) using Seaborn')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(8, 4))
sns.scatterplot(x=x, y=y, label='sin(x) points', marker='o',color='black')
plt.title('Scatter Plot of sin(x) using Seaborn')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.show()
```

# Distribution and Categorical Data Visualization

```python
# Load example dataset
tips = sns.load_dataset("tips")
```

```python
# Box Plot
plt.figure(figsize=(8, 4))
sns.boxplot(x='day', y='total_bill', data=tips,palette="Set2",hue='day')
plt.title('Box Plot of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
# Violin Plot
plt.figure(figsize=(8, 4))
sns.violinplot(x='day', y='total_bill', data=tips, palette="Set2",hue='day')
plt.title('Violin Plot of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
# Bar Plot
plt.figure(figsize=(8, 4))
sns.barplot(x='day', y='total_bill', data=tips, hue='day', palette="Set2", errorbar=None)
plt.title('Bar Plot of Average Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Average Total Bill')
plt.show()
```

```python
# Count Plot
plt.figure(figsize=(8, 4))
sns.countplot(x='day', data=tips, palette="Set2",hue='day')
plt.title('Count Plot of Days')
plt.xlabel('Day')
plt.ylabel('Count')
plt.show()
```

# Correlation Visualization with Heatmap

```python
# Compute correlation matrix
numeric_tips = tips.select_dtypes(include=np.number) # Select only numeric columns
corr = numeric_tips.corr() # Calculate correlation on numeric columns only
#-----------------------------------------------------------------------------
# Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="cividis", fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap of Tips Dataset')
plt.show()
```
